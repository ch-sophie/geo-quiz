```python
import pandas as pd
import requests
import json
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
API_KEY = os.getenv("API_KEY")
BASE_URL = "https://api.restcountries.com/countries/v5"
```

```python
headers = {"Authorization": f"Bearer {API_KEY}"}

response = requests.get(BASE_URL, headers=headers, params={"limit": 100})
print(response.status_code)

response = requests.get(BASE_URL, headers=headers, params={"limit": 100})
data = response.json()

countries = data["data"]["objects"]

df = pd.json_normalize(countries)
df.head()
```

```python
if "capitals" in df.columns:
    print("The 'capital' column exists!")
else:
    print("The 'capital' column is missing.")
```

```python
# Print the full dictionary of the first row
import pprint
pprint.pprint(df["capitals"].iloc[0])
```

```python
def extract_capital_name(capital_list):
    # Check if it's a valid list and has at least one item
    if isinstance(capital_list, list) and len(capital_list) > 0:
        first_item = capital_list[0]
        # Check if that item is a dictionary and contains the 'name' key
        if isinstance(first_item, dict):
            return first_item.get('name')
    return None # Return None if the data is missing or malformed

# Apply the function to create a clean column
df['capital_clean'] = df['capitals'].apply(extract_capital_name)
```

```python
print(df[['capital_clean']].head(10))
```

```python
#get lat and long of capitals 
df['capital_lat'] = df['capitals'].apply(lambda x: x[0].get('coordinates', {}).get('lat') if isinstance(x, list) and x else None)
df['capital_lng'] = df['capitals'].apply(lambda x: x[0].get('coordinates', {}).get('lng') if isinstance(x, list) and x else None)
```
